# Ejercicio final. Algoritmo k-NN para clasificación con NumPy

## Objetivo

Implementar desde cero un algoritmo **k-Nearest Neighbors (k-NN)** para clasificar una nueva fruta usando solamente NumPy.

## Idea del algoritmo

El algoritmo k-NN clasifica una nueva observación siguiendo estos pasos:

1. Recibe una observación nueva.
2. Calcula la distancia entre esa observación y todas las observaciones conocidas.
3. Selecciona los `k` vecinos más cercanos.
4. Revisa las etiquetas de esos vecinos.
5. Asigna la clase más frecuente entre los vecinos.

## Dataset

Usaremos el dataset de frutas del ejercicio anterior.

Cada fruta tiene tres características:

```text
[peso, diámetro, intensidad_color]
```

## Nueva fruta a clasificar

```python
nueva_fruta = np.array([155, 7.1, 8.2])
```

A simple vista, esta fruta parece cercana a las manzanas, pero el algoritmo debe decidirlo mediante distancias.

---

## Parte A. Cálculo de distancias

### Tareas

1. Crear la variable `nueva_fruta`.
2. Calcular la diferencia entre cada fila de `X` y `nueva_fruta`.
3. Elevar las diferencias al cuadrado.
4. Sumar por fila.
5. Aplicar raíz cuadrada para obtener la distancia euclidiana.
6. Verificar que el resultado tenga una distancia por cada muestra del dataset.

### Fórmula

Para dos puntos \(a\) y \(b\):

\[
d(a,b) = \sqrt{\sum (a_i - b_i)^2}
\]

````python
nueva_fruta = np.array([155, 7.1, 8.2])

# 1. Diferencias entre X y nueva_fruta
diferencias = X - nueva_fruta

# 2. Diferencias al cuadrado

diferencias_cuadrado = diferencias ** 2
# 3. Suma por fila
suma_filas = np.sum(diferencias_cuadrado, axis=1)

# 4. Raíz cuadrada: distancias euclidianas
distancias = np.sqrt(suma_filas)

# 5. Verifique la forma de las distancias
print("Forma de las distancias:", distancias.shape)
print("Distancias obtenidas para cada muestra:\n", distancias)
````

## Parte B. Selección de los k vecinos más cercanos

Ahora debe seleccionar los `k` vecinos más cercanos usando `np.argpartition`.

### Tareas

1. Definir `k = 3`.
2. Obtener los índices de los `k` vecinos más cercanos.
3. Mostrar las distancias de esos vecinos.
4. Mostrar las etiquetas de esos vecinos.
5. Ordenar los vecinos seleccionados de menor a mayor distancia para presentarlos mejor.

### Importante

Use `argpartition` para seleccionar los vecinos.  
Después puede usar `argsort` solo sobre el subconjunto seleccionado para ordenar la presentación.

````python
k = 3

# 1. Índices de los k vecinos más cercanos usando argpartition

indices_k = np.argpartition(distancias, k)[:k]
# 2. Distancias de los vecinos

distancias_vecinos = distancias[indices_k]
# 3. Etiquetas de los vecinos

etiquetas_vecinos = y[indices_k]
# 4. Ordenar los vecinos seleccionados por distancia
orden_relativo = np.argsort(distancias_vecinos)

indices_vecinos_ordenados = indices_k[orden_relativo]
distancias_vecinos_ordenados = distancias_vecinos[orden_relativo]
etiquetas_vecinos_ordenados = etiquetas_vecinos[orden_relativo]

print("Índices de los vecinos más cercanos:", indices_vecinos_ordenados)
print("Distancias correspondientes:", distancias_vecinos_ordenados)
print("Etiquetas correspondientes:", etiquetas_vecinos_ordenados)
````

## Parte C. Votación mayoritaria

Una vez seleccionados los vecinos, se debe determinar la clase más frecuente.

### Tareas

1. Usar `np.unique` con `return_counts=True` sobre las etiquetas de los vecinos.
2. Identificar la etiqueta con mayor frecuencia.
3. Mostrar la clase predicha para `nueva_fruta`.

### Pista

Puede usar:

```python
clases, conteos = np.unique(etiquetas_vecinos, return_counts=True)
```

Luego puede aplicar `np.argmax(conteos)` para encontrar la clase ganadora.

````python
# 1. Obtener clases y conteos
clases, conteos = np.unique(etiquetas_vecinos_ordenados, return_counts=True)

# 2. Índice de la clase ganadora
idx_ganadora = np.argmax(conteos)

# 3. Clase predicha
clase_predicha = clases[idx_ganadora]

# 4. Mostrar resultado final
print("Conteo de clases en el vecindario:", dict(zip(clases, conteos)))
print(f"Resultado final: La fruta con características {nueva_fruta} se clasifica como: {clase_predicha}")
````

## Parte D. Convertir el proceso en una función

Ahora implemente una función llamada `knn_predict`.

### Requisitos de la función

La función debe recibir:

```python
knn_predict(X_train, y_train, x_new, k=3)
```

Y debe retornar:

```python
clase_predicha
```

### Restricciones

- Debe usar NumPy.
- Debe usar `np.argpartition`.
- Debe usar distancia euclidiana.
- No debe usar librerías externas como scikit-learn.

````python
def knn_predict(X_train, y_train, x_new, k=3):
    """
    Clasifica una nueva observación usando k-NN con NumPy.

    Parámetros
    ----------
    X_train : np.ndarray
        Matriz de características de entrenamiento.
    y_train : np.ndarray
        Vector de etiquetas.
    x_new : np.ndarray
        Nueva observación a clasificar.
    k : int
        Número de vecinos más cercanos.

    Retorna
    -------
    str
        Clase predicha.
    """
    # 1. Calcular distancias
    diferencias = X_train - x_new
    diferencias_cuadrado = diferencias ** 2
    suma_filas = np.sum(diferencias_cuadrado, axis=1)
    distancias = np.sqrt(suma_filas)
    # 2. Seleccionar índices de los k vecinos más cercanos
    indices_k = np.argpartition(distancias, k)[:k]
    # 3. Obtener etiquetas de los vecinos
    etiquetas_vecinos = y_train[indices_k]
    # 4. Votación mayoritaria
    clases, conteos = np.unique(etiquetas_vecinos, return_counts=True)
    idx_ganadora = np.argmax(conteos)
    clase_predicha = clases[idx_ganadora]
    # 5. Retornar clase ganadora
    return clase_predicha
    pass


# Prueba de la función
nueva_fruta = np.array([155, 7.1, 8.2])
prediccion = knn_predict(X, y, nueva_fruta, k=3)
print("Predicción:", prediccion)
````
