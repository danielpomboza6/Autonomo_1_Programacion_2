## Ejercicio 3. Uso de `super()`

Vamos a mejorar el ejercicio anterior usando `super()`.

### Instrucciones
1. En la clase `Estudiante`, use `super().__init__(...)` para reutilizar el constructor de `Persona`.
2. Agregue un método para mostrar la carrera.

### Preguntas guía
- ¿Para qué sirve `super()`?
- ¿Qué ventaja tiene frente a volver a escribir los atributos heredados?

```python
# Ejercicio 3
# Reescriba aquí la clase Persona
class Persona():
  def __init__(self, nombre, edad):
    self.nombre = nombre
    self.edad = edad
  def mostrar_datos(self):
    print(f"Nombre: {self.nombre}")
    print(f"Edad: {self.edad}")

# Reescriba aquí la clase Estudiante usando super()

class Estudiante(Persona):
  def __init__(self,nombre, edad,carrera):
    super().__init__(nombre, edad)
    self.carrera = carrera
  def mostrar_datos(self):
    super().mostrar_datos()
    print(f"Carrera: {self.carrera}")
```

```python
from IPython.core.prefilter import MacroHandler
# Prueba del ejercicio 6
# Cree objetos de tipo Auto y Moto y muestre su información

class Vehiculo():
  def __init__(self, marca,modelo,color):
    self.marca = marca
    self.modelo = modelo
    self.color= color
  def mostrar_info(self):
    print(f"Marca: {self.marca}")
    print(f"Modelo: {self.modelo}")
    print(f"Color: {self.color}")

class Auto(Vehiculo):
  def __init__(self,marca,modelo,color,num_puertas):
    super().__init__(marca,modelo,color)
    self.num_puertas = num_puertas
  def mostrar_info(self):
    super().mostrar_info()
    print(f"Numero de puertas: {self.num_puertas}")

class Moto(Vehiculo):
  def __init__(self,marca,modelo,color):
    super().__init__(marca,modelo,color)
```

## Ejercicio 7. Sobrescritura de métodos

Vamos a redefinir un método heredado para que cada clase tenga un comportamiento particular.

### Instrucciones
1. En `Auto`, cree o redefina un método `encender()`.
2. En `Moto`, cree o redefina también el método `encender()`.
3. Compare los resultados.

### Preguntas guía
- ¿Qué diferencia hay entre heredar un método y sobrescribirlo?
- ¿En qué casos conviene sobrescribir?

```python
# Ejercicio 7
# Agregue o modifique aquí el método encender() en Auto y Moto

class Vehiculo():
  def __init__(self, marca,modelo,color):
    self.marca = marca
    self.modelo = modelo
    self.color= color
  def mostrar_info(self):
    print(f"Marca: {self.marca}")
    print(f"Modelo: {self.modelo}")
    print(f"Color: {self.color}")
  def encender(self):
    print("El vehiculo esta encendido")

class Auto(Vehiculo):
  def __init__(self,marca,modelo,color,num_puertas):
    super().__init__(marca,modelo,color)
    self.num_puertas = num_puertas
  def mostrar_info(self):
    super().mostrar_info()
    print(f"Numero de puertas: {self.num_puertas}")

  def encender(self):
    print(f"El auto {self.modelo} se encendio")

class Moto(Vehiculo):
  def __init__(self,marca,modelo,color):
    super().__init__(marca,modelo,color)
  def encender(self):
    print(f"La moto {self.modelo} se encendio")
```

```python
# Prueba del ejercicio 7
# Ejecute los métodos encender() en objetos de Auto y Moto

Auto_1=Auto("Toyota","Corolla","Rojo",4)


Auto_2=Auto("Peugeot","206",'Azul',2)


Moto_1=Moto("Honda","CBR","Negro")


mi_garaje=[Auto_1,Auto_2,Moto_1]

for vehiculo in mi_garaje:
  vehiculo.encender()
```

## Ejercicio 8. `super()` dentro de un método

No solo se puede usar `super()` en el constructor. También puede utilizarse dentro de otros métodos.

### Instrucciones
1. Cree una clase `Empleado` con un método `presentarse()`.

3. En `Docente`, redefina `presentarse()`.

### Preguntas guía

- ¿Qué parte se agrega en la clase hija?

```python
# Ejercicio 8
# Cree aquí la clase Empleado
class Empleado:
    def presentarse(self):
        print("Soy un empleado de la institución.")

# Cree aquí la clase Docente
class Docente(Empleado):
    def presentarse(self):
      super().presentarse()
      print("Mi rol específico es impartir clases como docente.")
```

## Ejercicio 10. Reto final

Desarrolle un sistema con:

- una clase base `Empleado`
- una clase hija `Docente`
- una clase hija `Administrativo`

### Requisitos
- usar herencia,
- usar `super()`,
- sobrescribir al menos un método,
- crear varios objetos,
- recorrerlos en una lista para demostrar polimorfismo.

### Sugerencia
Puede usar atributos como:
- nombre
- sueldo
- departamento
- materia

```python
# Ejercicio 10
# Desarrolle aquí su solución completa
class Empleado:
    def __init__(self, nombre, sueldo):
        self.nombre = nombre
        self.sueldo = sueldo

    def describir_rol(self):
        return f"Empleado: {self.nombre}, Sueldo: ${self.sueldo}"

# 2. Clase hija: Docente
class Docente(Empleado):
    def __init__(self, nombre, sueldo, materia):
        super().__init__(nombre, sueldo)
        self.materia = materia

    def describir_rol(self):
        detalle_padre = super().describir_rol()
        return f"{detalle_padre} | Materia: {self.materia} (Rol: Docente)"

# 3. Clase hija: Administrativo
class Administrativo(Empleado):
    def __init__(self, nombre, sueldo, departamento):
        super().__init__(nombre, sueldo)
        self.departamento = departamento

    def describir_rol(self):
        detalle_padre = super().describir_rol()
        return f"{detalle_padre} | Departamento: {self.departamento} (Rol: Administrativo)"

# 4. Crear varios objetos
empleado1 = Docente("Carlos Pérez", 1500, "Inteligencia Artificial")
empleado2 = Administrativo("Ana Gómez", 1200, "Recursos Humanos")
empleado3 = Docente("Luis Rivas", 1450, "Programación C++")

# 5. Lista para demostrar polimorfismo
lista_empleados = [empleado1, empleado2, empleado3]

print("--- Listado de Personal (Polimorfismo) ---")
for emp in lista_empleados:
    print(emp.describir_rol())
```
